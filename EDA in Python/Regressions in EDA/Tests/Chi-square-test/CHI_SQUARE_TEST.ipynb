<b>CHI SQUARE TEST</b>

<i>We use CHI SQUARE TEST to perform or figure out the relationship between two categorial features which are variate </i>

```python
dataset_table=pd.crosstab(dataset['sex'],dataset['smoker'])  
print(dataset_table)
```

```python
#Observed Values
Observed_Values=dataset_table.values
print("Observed Values:-\n",Observed_Values)
```

```python
val=stats.chi2_contingency(dataset_table)
```

```python
val
```

```python
Expected_Values=val[3]
```

```python
Expected_Values
```

```python
no_of_rows=len(dataset_table.iloc[0:2,0])
no_of_columns=len(dataset_table.iloc[0,0:2])
ddof=(no_of_rows-1)*(no_of_columns-1)    #degree of freedom
print("Degree of Freedom:",ddof)
alpha=0.05 #standardized val
```

<h4>$$
\chi^2 = \sum \frac{(o - e)^2}{e}
$$

*Chi-square formula*
</h4>

```python
from scipy.stats import chi2
chi_square=sum([(o-e)**2./e for o,e in zip(Observed_Values,Expected_Values)])
chi_square_statistic=chi_square[0]+chi_square[1]
```

```python
print("chi_square statistic:-",chi_square_statistic)
```

```python
critical_value=chi2.ppf(q=1-alpha,df=ddof)
print('critical_value:',critical_value)
```

<p align="center"><b><font size="4">P Value</font></b></p>


```python
p_value=1-chi2.cdf(x=chi_square_statistic,df=ddof)
print('p_value:',p_value)
print('Significance level:',alpha)
print("Degree of freedom",ddof)
```

<b>Wrto critical value , we find the relationship , and weather we will accept <br> or reject H0</b>

```python

if chi_square_statistic>=critical_value:  # chi_square value should be always greater than or equal to critical value
    print("Reject H0,There is a relationship between 2 categorical variables")
else:
    print("Retain H0,There is no relationship between 2 categorical variables")
    
```

<b>Wrto P-value , we find the relationship , and weather we will accept <br> or reject H0</b>

```python
if p_value<=alpha:
    print("Reject H0,There is a relationship between 2 categorical variables")
else:
    print("Retain H0,There is no relationship between 2 categorical variables")
```
