<h3>Performance metrics on MultiClass Classification Problems</h3>

```python
from sklearn import metrics
```

```python
C="Cat"
D="Dog"
F="Fox"
```

<h4>The precision for the cat class is the number of correctly predicted Cat out of all predicted cat!</h4>
<h4>the recall for Cat is the number of correctly predicted cat photos out of the number of actual cat</h4>



$$
\begin{array}{c|c|c}
  & \text{Predicted Positive (P)} & \text{Predicted Negative (N)} \\
  \hline
  \text{Actual Positive (P)} & \text{True Positives (TP)} & \text{False Negatives (FN)} \\
  \hline
  \text{Actual Negative (N)} & \text{False Positives (FP)} & \text{True Negatives (TN)} \\
\end{array}
$$


```python
 #True values
y_true=[C,C,C,C,C,C, F,F,F,F,F,F,F,F,F,F, D,D,D,D,D,D,D,D,D]
 #Predicted values
y_pred=[C,C,C,C,D,F, C,C,C,C,C,C,D,D,F,F, C,C,C,D,D,D,D,D,D]

#Print the conusion matrix
print(metrics.confusion_matrix(y_true,y_pred))

#print the precision and recall , among the other metrics
print(metrics.classification_report(y_true,y_pred,digits=3))
```

<p>Precision = Of all the people I called “my friend,” how many actually are?

Recall = Of all my real friends, how many did I correctly call “my friend”?</p>

\[
\begin{array}{|c|c|}
\hline
\textbf{Metric} & \textbf{Formula} \\
\hline
\text{Precision} & \dfrac{TP}{TP + FP} \\
\hline
\text{Recall} & \dfrac{TP}{TP + FN} \\
\hline
\end{array}
\]
