<h2>missing Values</h2>

```python

## Let us capture all the nan values
## First lets handle Categorical features which are missing
features_nan=[feature for feature in dataset.columns if dataset[feature].isnull().sum()>1 and dataset[feature].dtypes=='O']

for feature in features_nan:
    print("{}: {}% missing values".format(feature,np.round(dataset[feature].isnull().mean(),4)))
```

```python

## Replace missing value with a new label
def replace_cat_feature(dataset,features_nan):
    data=dataset.copy()
    data[features_nan]=data[features_nan].fillna('Missing')   ## Jaha jaha bhi 0 hoga , it would be replaced bt "missing"
    return data

dataset=replace_cat_feature(dataset,features_nan)

dataset[features_nan].isnull().sum()
```

```python
dataset.head()
```

```python
# dataset.drop(dataset['LowQualFinSF'])
```

```python
## Now lets check for numerical variables the contains missing values
numerical_with_nan=[feature for feature in dataset.columns if dataset[feature].isnull().sum()>1 and dataset[feature].dtypes!='O']# !=(not equal)

## We will print the numerical nan variables and percentage of missing values

for feature in numerical_with_nan:
    print("{}: {}% missing value".format(feature,np.around(dataset[feature].isnull().mean(),4)))
```

```python

## Replacing the numerical Missing Values

for feature in numerical_with_nan:
    ## We will replace by using median since there are outliers
    median_value=dataset[feature].median()
    
    ## create a new feature to capture nan values
    dataset[feature+'nan']=np.where(dataset[feature].isnull(),1,0)
    dataset[feature].fillna(median_value,inplace=True)
    
dataset[numerical_with_nan].isnull().sum()
    
```

```python
dataset.head(50)
```

```python
## Temporal Variables (Date Time Variables)  ##The date time variables

for feature in ['YearBuilt','YearRemodAdd','GarageYrBlt']:
       
    dataset[feature]=dataset['YrSold']-dataset[feature]
```

<h3>Numerical Variables</h3>
Since the numerical variables are skewed we will perform log normal distribution

```python
import numpy as np
num_features=['LotFrontage','LotArea','1stFlrSF','GrLivArea','SalePrice']

for feature in num_features:
    dataset[feature]=np.log(dataset[feature])
```

```python
dataset.head() # The selected features were converted into their log normal distribution form!
```

<h3>Handling Rare Categorial features</h3>
<p>We will now remove categorial variables that are presebt in less than 1% of the observations</p>

```python
categorial_features=[feature for feature in dataset.columns if dataset[feature].dtypes=='O']
```

```python
categorial_features
```

```python
for feature in categorial_features:
    temp=dataset.groupby(feature)['SalePrice'].count()/len(dataset)
    temp_df=temp[temp>0.01].index
    dataset[feature]=np.where(dataset[feature].isin(temp_df),dataset[feature],'Rare_var')
```

```python
dataset.head(100)
```

<h3>Feature Scaling</h3>

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Features to scale (exclude Id and SalePrice)
feature_scale = [f for f in dataset.columns if f not in ['Id', 'SalePrice']]

# Separate numeric and categorical columns
numeric_features = dataset[feature_scale].select_dtypes(include=['int64', 'float64']).columns
categorical_features = [f for f in feature_scale if f not in numeric_features]

# Initialize and fit scaler
scaler = MinMaxScaler()
scaler.fit(dataset[numeric_features])

# Transform numeric features
scaled_numeric = pd.DataFrame(scaler.transform(dataset[numeric_features]), columns=numeric_features)

# Combine Id, SalePrice, scaled numeric, and categorical columns
data = pd.concat([
    dataset[['Id', 'SalePrice']].reset_index(drop=True),
    scaled_numeric.reset_index(drop=True),
    dataset[categorical_features].reset_index(drop=True)
], axis=1)
```
