```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Lasso
from sklearn.feature_selection import SelectFromModel

pd.pandas.set_option('display.max_columns',None)
```

```python
dataset=pd.read_csv('X_train.csv')
```

```python
dataset.head()
```

```python
##Capture the dependent feature
y_train=dataset['SalePrice']
```

```python
X_train=dataset.drop(['Id','SalePrice'],axis=1)
```

```python
### Apply Feature Selection
# first, I specify the Lasso Regression model, and I
# select a suitable alpha (equivalent of penalty).
# The bigger the alpha the less features that will be selected.

# Then I use the selectFromModel object from sklearn, which
# will select the features which coefficients are non-zero

import pandas as pd
from sklearn.linear_model import Lasso
from sklearn.feature_selection import SelectFromModel

# Assuming X_train contains features and y_train is SalePrice
# One-hot encode categorical columns
X_train_encoded = pd.get_dummies(X_train, drop_first=True)

# Initialize Lasso for feature selection
feature_sel_model = SelectFromModel(Lasso(alpha=0.005, random_state=0))
feature_sel_model.fit(X_train_encoded, y_train)

# Get selected features
selected_features = X_train_encoded.columns[feature_sel_model.get_support()]
print("Selected features:\n", selected_features)

```

```python
feature_sel_model.get_support()
```

```python
import numpy as np

# If ne-hot encoded  X_train
X_train_used = X_train_encoded  # the numeric-only version used for Lasso

# Get the selected features
selected_feat = X_train_used.columns[feature_sel_model.get_support()]

# Print stats
print('Total features: {}'.format(X_train_used.shape[1]))
print('Selected features: {}'.format(len(selected_feat)))
print('Features with coefficients shrank to zero: {}'.format(
    np.sum(feature_sel_model.estimator_.coef_ == 0)
))
```
