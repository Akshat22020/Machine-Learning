<h3>Bank note Authentication</h3>
<p>Data were extracted from images that were taken from genuine and forged banknote-like specimens. For digitization, an industrial camera usually used for print inspection was used.The final images have 400x 400 pixels. Due to the object lens and distance to the investigated object gray-scale pictures with a resolution of about 660 dpi were gained. Wavelet Transform tool were used to extract features from images. The attributes are extracted attributes of each image.</p>

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("dataset.csv")
```

```python
df.head()
```

```python
##Dependent and Independent Features division (as always)
X=df.iloc[:,:-1]  ## X contains all the values on the each of the rows but not the last column
y=df.iloc[:,-1]   ##all rows  and all columns except last 
```

```python
X.head()
```

```python
y.head()
```

```python
##Test train split
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=0)
```

```python
##implimenting random forest classifier
from sklearn.ensemble import RandomForestClassifier
classifier=RandomForestClassifier()
classifier.fit(X_train,y_train)
```

```python
## prediction
y_pred=classifier.predict(X_test)
```

```python
##check accuracy
from sklearn.metrics import accuracy_score
score = accuracy_score(y_test,y_pred)
```

```python
score
```

```python
##create a pickle file using serialization
import pickle as pkl
pkl_out = open("classifier.pkl","wb")  ##creationb of pickle file
pkl.dump(classifier,pkl_out)  ## very important to dump 
pkl_out.close()



##The pickle.dump function is used in Python to serialize Python objects into a byte stream, 
##allowing for the storage or transmission of these objects
```
