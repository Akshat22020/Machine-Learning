<h1 style.bgcolor="blue">Corona virus live tracking</h1>

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_json('https://covid19dashboard.mohfw.gov.in/data/datanew.json')
df
```

```python
df=df.iloc[:-1,1:]
```

```python
df.head()
```

```python
df.tail()
```

```python
import numpy as np
total_states=np.arange(len(df['state_name']))
```

```python
total_states
```

```python
df['active']
```

```python
max(df['positive'])
```

```python
from matplotlib.pyplot import figure
```

<h3 text_align='center'> Total Positive Cases of states in india </h3>

```python
figure(num=None,figsize=(9,6),dpi=80,facecolor='w',edgecolor='k')
plt.barh(total_states,df['positive'],align='center',alpha=0.5,color=(1,0,0),edgecolor=(0.5,0.2,0.8))
plt.yticks(total_states,df['state_name'])
plt.xlim(1,max(df['positive'])+10)
plt.xlabel('Positive Number of cases')
plt.title('Corona Virus Cases')
plt.show()
```

<h4><i>Active number of cases</i></h4>

```python
#from matplotlib.pyplot import figure
figure(num=None,figsize=(9,6),dpi=80,facecolor='w',edgecolor='k')
plt.barh(total_states,df['active'],align='center',alpha=0.5,color=(1,0.5,0),edgecolor=(0.5,0.4,0.8))
plt.yticks(total_states,df['state_name'])
plt.xlim(1,max(df['active'])+10)
plt.xlabel('Active Number of cases')
plt.title('Corona Virus Cases')
plt.show()
```

```python
figure(num=None,figsize=(9,6),dpi=80,facecolor='w',edgecolor='k')
plt.barh(total_states,df['death'],align='center',alpha=0.5,color=(1,0.5,0),edgecolor=(0.5,0.4,0.8))
plt.yticks(total_states,df['state_name'])
plt.xlim(1,max(df['death'])+10)
plt.xlabel('Number of deaths')
plt.title('Corona Virus Cases')
plt.show()
```
